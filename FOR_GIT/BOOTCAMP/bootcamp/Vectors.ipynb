```python
# Creating a more versatile function that can handle any number of vectors and coefficients
def linear_combination(vectors, coefficients):
    # check if the number of vectors matches the number of coefficients
    num_dimensions = len(vectors[0])
    for vector in vectors:
        if len(vector) != num_dimensions:
            print("All vectors must have the same dimension.")
            break

    # Initialize a result vector with zeros, assuming all vectors have the same dimension
    result = [0] * num_dimensions
    
    for i in range(num_dimensions):
        for j,vector in enumerate(vectors):
            result[i] +=  coefficients[j] * vector[i]
    
    return tuple(result)
```

```python
v1=(3,2,5)
v2=(4,6,1)
coefficients = [2, 3]

result = linear_combination([v1, v2], coefficients)
result
```
