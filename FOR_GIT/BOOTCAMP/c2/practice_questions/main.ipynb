```python
#Read an integer and print whether it is even or odd.
n=int(input("Enter an integer: "))
if n%2==0:
    print("EVEN")
else:
    print("ODD")
```

```python
#Read two integers and print the largest one.
num1=int(input("Enter first integer: "))
num2=int(input("Enter second integer: "))
if num1>num2:
    print(f"{num1} is larger than {num2}")
else:
    print(f"{num2} is the larger than {num1}")
```

```python
#Write a program to read an age and classify it as Child (<13), Teen (13–17), or Adult (18+). If age is negative, print "Invalid".

age=int(input("Enter your age: "))
if age<0:
    print("Invalid")
elif age<13:
    print("Child")
elif age<18:
    print("Teen")
else:
    print("Adult")
```

```python
#Write a program to compare three sides of a triangle and print whether the triangle is valid or not. A triangle is valid if the sum of any two sides is greater than the third side.

s1=int(input("Enter first side of triangle: "))
s2=int(input("Enter second side of triangle: "))
s3=int(input("Enter third side of triangle: "))

if s1+s2>s3 and s1+s3>s2 and s2+s3>s1:
    print("Valid triangle")
else:
    print("Invalid triangle")
```

```python
"""
Write a program to determine if a given year is a leap year. A year is a leap year if:
It is divisible by 4 and not divisible by 100, or
It is divisible by 400

"""

year=int(input("Enter a year: "))
if (year%4==0 and year%100!=0) or (year%400==0):
    print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
```

```python
#Write a program to calculate the sum of all even numbers from 1 to N (inclusive).

N=int(input("Enter a number: "))
sum_even=0
for i in range(2, N+1, 2):
    sum_even += i
print(f"The sum of all even numbers from 1 to {N} is {sum_even}")
```
