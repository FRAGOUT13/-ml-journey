```python
#find longest consecutive sequence of identical digits.

n = int(input())

temp = -1
c = 0
seq = 0

while n > 0:
    d = n % 10

    if d == temp:
        c += 1
    else:
        c = 1

    if c > seq:
        seq = c

    temp = d
    n = n // 10

print(seq)
```
