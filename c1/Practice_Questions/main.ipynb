```python
#Write a Python program to take two integers x and y as input from a user and solve the following expression: (x^y+(x+y)^(x-y)) 

x=int(input("enter the value of x: "))
y=int(input("enter the value of y: "))
final=(x**y)+(x+y)**(x-y)
print(final)
```

```python
#Write a Python program to take a temperature in degrees Celsius as input from a user and convert it to the Fahrenheit scale.

temperature=float(input("enter the temperature in degrees Celsius: "))
fahrenheit=(temperature*9/5)+32
print(fahrenheit)
```

```python
#You are given a base salary and a bonus amount. Update the salary by adding the bonus using an assignment operator. Then, check if the updated salary is more than 50000 and the bonus is more than 5000.

base_salary=float(input("enter the base salary: "))
bonus=float(input("enter the bonus amount: "))
base_salary+=bonus
if base_salary>50000 and bonus>5000:
    print("The updated salary is more than 50000 and the bonus is more than 5000.")
else:
    print("The conditions are not met.")
```

```python
#Write a program that takes two variables representing a person's first name and last name, and prints the full name in the format "Last, First".

first_name=input("enter the first name: ")
last_name=input("enter the last name: ")
print(f"{last_name}, {first_name}")
```

```python
#Given the principal amount, rate of interest, and time in years, calculate and print the simple interest using the formula SI = (P * R * T) / 100.

P=float(input("enter the principal amount: "))
R=float(input("enter the rate of interest: "))
T=float(input("enter the time in years: "))
SI=(P * R * T) / 100
print(SI)
```

```python
#Write a program that takes hours, minutes, and seconds and prints the total time in seconds.

hours=int(input("enter the hours: "))
minutes=int(input("enter the minutes: "))
seconds=int(input("enter the seconds: "))
total_seconds=hours*3600+minutes*60+seconds
print(total_seconds)
```

```python
#Given two variables a and b, swap their values and print them.

a=int(input("enter the value of a: "))
b=int(input("enter the value of b: "))
a, b = b, a
print(f"The value of a is: {a}")
print(f"The value of b is: {b}")
```

```python
#Given the base price of a product and a GST rate (in percent), compute the final price.

base_price=float(input("enter the base price of the product: "))
gst_rate=float(input("enter the GST rate (in percent): "))
final_price=base_price+(base_price*gst_rate/100)
print(final_price)
```

```python
#You are given three boolean inputs a, b, and c (entered as 0 for False and 1 for True). Compute and print the result of the expression:(a AND NOT b) OR (b AND NOT c)  

a=int(input("enter the value of a (0 for False, 1 for True): "))
b=int(input("enter the value of b (0 for False, 1 for True): "))
c=int(input("enter the value of c (0 for False, 1 for True): "))
result=(a and not b) or (b and not c)
print(result)
```
