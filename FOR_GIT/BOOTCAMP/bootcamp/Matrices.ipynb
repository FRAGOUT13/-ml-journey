Matrices in Python

```python
# declaring matrices as list of lists or nested lists

A= [[1,2],
     [3,4]]

B=[[5,6],
     [7,8]]


#addition

addition=[[A[i][j]+B[i][j] for j in range(len(A[0]))] for i in range(len(A))]
print(addition)

#subtraction

subtraction=[[A[i][j]-B[i][j] for j in range(len(A[0]))] for i in range(len(A))]
print(subtraction)

#scalar multiplication
scalar=2
scalar_multiplication=[[scalar*A[i][j] for j in range(len(A[0]))] for i in range(len(A))]
print(scalar_multiplication)

# transpose
transpose=[[A[j][i] for j in range(len(A))] for i in range(len(A[0]))]
print(transpose)
```

```python
#Matrix multiplication

#Matrix Vector multiplication
A=[[1,2],
   [3,4]]

v=[5,6]

multiplication=[sum(A[i][j] * v[j] for j in range(len(v))) for i in range(len(A))]
print(multiplication)
print()


#Matrix Matrix multiplication
A=[[1,2],
   [3,4]]

B=[[5,6],
   [7,8]]

multiplication=[[sum(A[i][k] * B[k][j] for k in range(len(B))) for j in range(len(B[0]))] for i in range(len(A))]
print(multiplication)
```

```python
#Determinant of a 2x2 matrix
A=[[1,2],
   [3,4]]

def determinant(matrix):
    return matrix[0][0]*matrix[1][1]-matrix[0][1]*matrix[1][0]
print(determinant(A))
print()


#Inverse of a 2x2 matrix
B=[[1,2],
   [3,4]]

def inverse_2x2(matrix):
    det = determinant(matrix)
    if det == 0:
        raise ValueError("Matrix is singular and cannot be inverted.")
    inverse_matrix = [[matrix[1][1]/det, -matrix[0][1]/det],
                      [-matrix[1][0]/det, matrix[0][0]/det]]
    return inverse_matrix

print(inverse_2x2(B))
print()
```
