```python
# Write a function that takes a number and returns whether it is even or odd.

def evenodd(num):
    if(num%2==0):
        return "Even"
    else:
        return "Odd"

evenodd(5)
```

```python
#Write a function factorial(n) that returns the factorial of a number using a loop.

def facto(n):
    result=1
    for i in range(1,n+1):
        result*=i
    return result
facto(3)
```

```python
# Write a lambda that takes two numbers and returns the larger one.

large=lambda a,b: a if a>b else b
print(large(8,7))
```

```python
# Write a lambda that takes a number and returns its cube.

cub=lambda x: x*x*x
print(cub(4))
```
