```python
"""
You are given four integers a, b, c, and d. Perform the following operations:



1. Update a using the formula a = a * b + c

2. Check if a is divisible by d and greater than both b and c

3. Print the result of this condition (True or False)
"""

a=int(input("Enter a number: "))
b=int(input("Enter another number: "))
c=int(input("Enter a third number: "))
d=int(input("Enter a fourth number: "))

a=a*b+c

if a%d==0 and b<a>c:
    print("True")
else:
    print("False")
```

```python
"""
You are given a 3-digit number. 
Extract the first and last digits, 
then form a new 2-digit number where the first digit is 
the last digit of the input and the second digit is the first digit of the input.

"""

num=int(input("Enter a 3-digit number: "))
first_digit=num//100
last_digit=num%10
new_number=last_digit*10+first_digit
print(new_number)
```

```python
#You are given a total number of seconds. Convert it to hours, minutes, and seconds and print it in the format hh:mm:ss.

total_seconds = int(input("Enter the total number of seconds: "))
hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60
print(f"{hours:02d}:{minutes:02d}:{seconds:02d}")
```

```python
"""

A vehicle travels two different segments of a journey:



1. It covers the first segment at speed s1 for time t1

2. Then it covers the second segment at speed s2 for time t2



Calculate and print the overall average speed.

Note: The average speed should be a whole number


"""

speed1 = int(input("Enter the speed for the first segment (in km/h): "))
time1 = int(input("Enter the time for the first segment (in hours): "))
speed2 = int(input("Enter the speed for the second segment (in km/h): "))
time2 = int(input("Enter the time for the second segment (in hours): "))
average_speed = (speed1 * time1 + speed2 * time2) / (time1 + time2)
print(int(average_speed))
```

```python
"""

You are given an amount in rupees. 
Calculate how many 100s, 50s, 10s, and remaining rupees it contains using successive modulus and floor division.

"""

amount = int(input("Enter the amount in rupees: "))
hundreds = amount // 100
amount %= 100
fifties = amount // 50
amount %= 50
tens = amount // 10
amount %= 10
print(f"100s: {hundreds}, 50s: {fifties}, 10s: {tens}, Remaining: {amount}")
```
